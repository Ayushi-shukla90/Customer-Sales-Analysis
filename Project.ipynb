```python
import pandas as pd
from sqlalchemy import create_engine

# =========================
# Load CSV
# =========================

df = pd.read_csv(r"C:\Users\netware\Downloads\Project 1(Sheet3).csv")

# =========================
# Data Cleaning
# =========================

# Fill missing values
df['Review Rating'] = df['Review Rating'].fillna(
    df['Review Rating'].mean()
)

# Clean column names
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Rename column
df.rename(
    columns={'purchase_amount_(usd)': 'purchase_amount'},
    inplace=True
)

# Create Age Groups
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']

df['age_groups'] = pd.qcut(
    df['age'],
    q=4,
    labels=labels
)

# Clean purchase frequency column
df['frequency_of_purchases'] = (
    df['frequency_of_purchases']
    .str.strip()
    .str.lower()
)

# Mapping
frequency_mapping = {
    'fortnightly': 14,
    'weekly': 7,
    'monthly': 30,
    'quarterly': 90,
    'biweekly': 14,
    'annually': 365,
    'every 3 months': 90
}

# Create numeric column
df['purchase_frequency_days'] = (
    df['frequency_of_purchases']
    .map(frequency_mapping)
)

# Convert datatype
df['purchase_frequency_days'] = (
    df['purchase_frequency_days']
    .astype('Int64')
)

# Drop unnecessary column
df = df.drop('promo_code_used', axis=1)

# MySQL Connection
username = "root"
password = "netware"
host = "127.0.0.1"
port = 3306
database = "customer_data"


# Connection string
connection_string = (
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

# Create engine
engine = create_engine(connection_string)


df.to_sql(
    name='customer_shopping_data',
    con=engine,
    if_exists='replace',
    index=False
)

print("Data uploaded successfully to MySQL")
```
